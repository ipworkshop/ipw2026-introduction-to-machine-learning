### Ce este un Perceptron?

Un perceptron este cel mai simplu tip de neuron artificial. Gândește-te la el ca la o mașinărie minusculă care ia decizii de tip "Da" sau "Nu" (1 sau 0).

Să zicem că perceptronul trebuie să decidă dacă ieși afara.

* **Intrările (Inputs/Fetures):** Sunt informațiile pe care le primește. De exemplu: "Plouă afară?" și "Ai timp liber?".
* **Ponderile (Weights):** Cât de importantă este fiecare informație pentru tine. Poate urăști ploaia (pondere negativă mare), dar ai foarte mult timp liber (pondere pozitivă).
* **Bias-ul (Pragul):** Lenea ta naturală. Cât de greu este să te ridici din pat în general, indiferent de vreme.

Perceptronul adună toate aceste informații, le cântărește, adună bias-ul, iar dacă rezultatul depășește un anumit prag, decizia este 1 ("Ies la plimbare!"). Altfel, este 0 ("Stau în casă!").

---

### Reprezentarea matematică

Pentru a pune neuronul să calculeze, traducem exemplul de mai sus într-o formulă matematică.

Avem următoarele elemente:

* $A$: Datele de intrare (informațiile noastre).
* $W$: Ponderile (importanța fiecărei informații).
* $b$: Bias-ul (pragul de bază).

Perceptronul funcționează în doi pași simpli:

**1. Combinația liniară**
Neuronul înmulțește fiecare intrare cu ponderea ei și adună bias-ul.


$$z = A \cdot W + b$$

**2. Funcția de activare**
Neuronul ia rezultatul $z$ și îl trece printr-o funcție pentru a lua decizia finală, $\hat{y}$. Cel mai des folosim o funcție treaptă (Step Function): dacă rezultatul e pozitiv, zice Da (1). Dacă e negativ, zice Nu (0).


$$\hat{y} = \begin{cases}  1 & \text{dacă } z \geq 0 \\ 0 & \text{dacă } z < 0  \end{cases}$$

### De la o decizie la mai multe:

### Cum știe modelul dacă a greșit? Funcția de Cost (Loss)

Am obținut niște predicții, dar modelul abia s-a "născut". Ponderile ($W$) cu care am început au fost alese la întâmplare, deci probabil a ghicit prost.

Ca să poată învăța, modelul are nevoie de o notă. Trebuie să îi spunem matematic cât de tare a dat greș. Această "notă" se numește **Funcție de Cost (Loss Function)**.

* Dacă eroarea (Loss) este mare $\rightarrow$ modelul a greșit grav.
* Dacă eroarea (Loss) este aproape 0 $\rightarrow$ modelul este perfect.

Scopul întregului algoritm de Machine Learning este **să minimizeze acest Loss**.

### Cum arată realitatea? (One-Hot Encoding)

Pentru a calcula greșeala, trebuie să comparăm predicția modelului cu realitatea (răspunsul corect din baza de date).

Să ne amintim clasele noastre: 0 = Pisică, 1 = Câine, 2 = Pasăre.
Dacă poza numărul 1 este cu adevărat un Câine, nu putem scrie pur și simplu "1". O transformăm într-un vector care arată 100% sigur pe poziția Câinelui și 0 pe restul. Această tehnică se numește **One-Hot Encoding**:

* Adevărul pentru Câine: `[0, 1, 0]`

Dacă modelul a prezis `[0.1, 0.8, 0.1]`, eroarea va fi mică, pentru că 0.8 este aproape de 1.
Dacă modelul a prezis `[0.9, 0.05, 0.05]`, eroarea trebuie să fie uriașă, pentru că a fost sigur că e Pisică (0.9), când de fapt era Câine.

### Cross-Entropy (Entropia Încrucișată)

O funcție matematică pentru a măsura distanța dintre două probabilități este **Cross-Entropy**.

Ea funcționează astfel: se uită *doar* la probabilitatea pe care modelul a dat-o clasei corecte. Apoi, aplică un **logaritm** (mai exact, logaritm natural). Logaritmul ajuta prin prorpietatiile sale: pedepsește foarte, foarte dur un model care este "extrem de sigur, dar greșește".

Formula matematică pe un singur exemplu este:


$$\text{Loss} = - \sum_{i=1}^{K} Y_i \cdot \log(P_i)$$

Unde:

* $Y_i$ este valoarea reală (1 pentru clasa corectă, 0 pentru restul).
* $P_i$ este probabilitatea prezisă de model.

Fiindcă în $Y$ avem un singur 1 și restul sunt 0, formula calculează practic doar: `-log(probabilitatea_clasei_corecte)`.

### Coborârea pe Gradient (Gradient Descent): Cum învățăm din greșeli?

Avem eroarea (Loss-ul). Acum ce facem cu ea? Cum modificăm ponderile $W$ și bias-ul $b$ ca data viitoare să greșească mai puțin?

În matematică, această "pantă" se numește **Gradient**. Gradientul ne arată direcția în care eroarea *crește*. Noi vrem să minimizăm eroarea, așa că vom face un pas în direcția *opusă* gradientului.

**Secretul magic al matematicii:**
Datorită modului inteligent în care am ales funcțiile (Softmax + Cross-Entropy), derivata lor combinată se simplifică la o formulă incredibil de elegantă.
"Panta" sau eroarea directă la nivelul predicției este pur și simplu diferența dintre ce am prezis și ce trebuia să fie:


$$\text{Eroare} = P - Y$$

1. Calculăm gradientul pentru ponderi: $grad\_W = \frac{1}{N} A^T \cdot (P - Y)$
