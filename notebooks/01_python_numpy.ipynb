## 3. Liste

O listă este o structură care păstrează mai multe valori într-o anumită ordine.

Fiecare element are o poziție, numită **index**, prin care îl putem accesa. În Python, indexarea începe de la `0`, deci primul element are indexul `0`, al doilea are indexul `1` și așa mai departe.

Listele pot fi modificate după creare: putem adăuga, elimina sau înlocui elemente.

````python
scores = [82, 71, 95, 64, 88]

print("Lista completă:", scores)
print("Primul scor:", scores[0])
print("Ultimul scor:", scores[-1])
print("Primele trei scoruri:", scores[:3])
````

## 6. De la liste la array-uri

## 7. Vectori și operații element-wise

Un vector este un array NumPy cu o singură dimensiune, folosit pentru a păstra o colecție de valori numerice, cum ar fi scorurile tuturor studenților.

Când aplicăm o operație aritmetică unui vector, NumPy aplică aceeași operație fiecărui element. Astfel, putem transforma toate valorile dintr-o singură instrucțiune, fără să scriem manual o buclă `for`.

````python
scores = np.array([82, 64, 95, 76])

print("Bonus de 5 puncte:", scores + 5)
print("Scoruri înjumătățite:", scores / 2)
print("Scoruri la pătrat:", scores ** 2)
````

## 8. Matrice

O matrice este un array NumPy cu două dimensiuni, organizat pe linii și coloane.

Într-un dataset de Machine Learning:

- fiecare **linie** reprezintă un exemplu separat, de exemplu un student;
- fiecare **coloană** reprezintă o caracteristică măsurată pentru toate exemplele, de exemplu numărul de ore studiate sau numărul de exerciții rezolvate.

Astfel, o matrice ne permite să păstrăm mai multe exemple și mai multe caracteristici într-o singură structură numerică.

## 11. Produsul scalar

Avem un vector de caracteristici:

```text
[hours_studied, exercises_solved]
```

și un vector de ponderi:

```text
[importance_hours, importance_exercises]
```

Fiecare pondere arată cât de importantă este caracteristica asociată.

Produsul scalar combină caracteristicile și ponderile într-o singură valoare numerică. În Machine Learning, acest tip de calcul este folosit frecvent pentru a obține un scor sau o predicție.

````python
features = np.array([4, 18])
weights = np.array([0.7, 0.3])

manual_score = (
    features[0] * weights[0]
    + features[1] * weights[1]
)

numpy_score = np.dot(features, weights)

print("Calcul manual:", manual_score)
print("Cu NumPy:", numpy_score)
````
