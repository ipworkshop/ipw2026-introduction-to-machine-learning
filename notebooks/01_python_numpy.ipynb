## 9. Matrice

O matrice este un array NumPy cu două dimensiuni, organizat pe linii și coloane.

Într-un dataset de Machine Learning:

- fiecare **linie** reprezintă un exemplu separat, de exemplu un student;

## 11. Broadcasting și centrarea datelor

Până acum am calculat media fiecărei caracteristici. Următorul pas este să vedem cât de mult se abate fiecare valoare față de media coloanei sale.

Pentru asta, scădem vectorul mediilor din fiecare linie a matricei. Deși matricea are mai multe linii, NumPy aplică automat același vector tuturor liniilor. Acest comportament se numește **broadcasting**.

Rezultatul se numește **date centrate**: după scădere, media fiecărei coloane devine aproximativ `0`.

`````python
X = np.array([
    [4, 18],
    [2, 10],
    [5, 25],
    [3, 14],
], dtype=float)

feature_means = X.mean(axis=0)
centered_X = X - feature_means

print("Medii:", feature_means)
print("Date centrate:")
print(centered_X)
print("Noile medii:", centered_X.mean(axis=0))
`````

## 12. Produsul scalar

Avem un vector de caracteristici:

```text
[hours_studied, exercises_solved]
````

și un vector de ponderi:

```text
[importance_hours, importance_exercises]
```

Fiecare pondere arată cât de importantă este caracteristica asociată.

Produsul scalar combină caracteristicile și ponderile într-o singură valoare numerică. În Machine Learning, acest tip de calcul este folosit frecvent pentru a obține un scor, o predicție sau intrarea unui neuron artificial.



`````python
features = np.array([4, 18])
weights = np.array([0.7, 0.3])

manual_score = (
    features[0] * weights[0]
    + features[1] * weights[1]
)

numpy_score = np.dot(features, weights)

print("Calcul manual:", manual_score)
print("NumPy:", numpy_score)
`````
