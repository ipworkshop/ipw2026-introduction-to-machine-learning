## 6. De la liste la array-uri

Până acum am folosit liste Python pentru a organiza datele. Pentru Machine Learning, însă, lucrăm frecvent cu multe valori numerice și vrem să aplicăm aceeași operație asupra tuturor valorilor.

NumPy oferă structura `ndarray`, un array multidimensional optimizat pentru calcule numerice.

Un array NumPy păstrează, de regulă, valori de același tip. Datorită acestei organizări, datele pot fi păstrate compact în memorie și procesate eficient.

O mare parte din operațiile NumPy sunt executate de un engine intern implementat în cod compilat, în special C. Atunci când scriem o operație precum:

`numpy_scores + 5`

nu mai scriem noi o buclă Python care trece prin fiecare scor. NumPy trimite operația către engine-ul său optimizat, care procesează întregul array.

Această abordare se numește **vectorizare**.

Vectorizarea nu înseamnă că elementele nu mai sunt procesate individual. Înseamnă că bucla este executată în cod optimizat, nu de interpretorul Python.

Exemplu:

```python
python_scores = [82, 64, 95, 76]
python_bonus = [score + 5 for score in python_scores]

numpy_scores = np.array([82, 64, 95, 76])
numpy_bonus = numpy_scores + 5
```

Ambele variante produc același rezultat, dar varianta NumPy exprimă direct operația matematică aplicată întregului vector.

Acest stil apare foarte des în Machine Learning, unde lucrăm cu vectori, matrici și batch-uri de date.

````python
python_scores = [82, 64, 95, 76]
numpy_scores = np.array([82, 64, 95, 76])

print(type(python_scores))
print(type(numpy_scores))
print("Shape:", numpy_scores.shape)
print("Tipul elementelor:", numpy_scores.dtype)
````

## 8. Matrice

O matrice este un array NumPy cu două dimensiuni, organizat pe linii și coloane.

Într-un dataset de Machine Learning:

- fiecare **linie** reprezintă un exemplu separat, de exemplu un student;
- fiecare **coloană** reprezintă o caracteristică măsurată pentru toate exemplele, de exemplu numărul de ore studiate sau numărul de exerciții rezolvate.

Astfel, o matrice ne permite să păstrăm mai multe exemple și mai multe caracteristici într-o singură structură numerică.

````python
X = np.array([
    [4, 18],
    [2, 10],
    [5, 25],
    [3, 14],
])

print(X)
print("Shape:", X.shape)
print("Număr de dimensiuni:", X.ndim)
````

În matricea `X`:

- coloana `0` reprezintă `hours_studied`;
- coloana `1` reprezintă `exercises_solved`.

Putem selecta o linie, o coloană sau o porțiune din matrice folosind indecși.

````python
print("Primul student:", X[0])
print("Orele tuturor studenților:", X[:, 0])
print("Exercițiile tuturor studenților:", X[:, 1])
print("Primele două exemple:")
print(X[:2])
````

### Adăugarea unei axe: `np.expand_dims`

Un singur student din matrice are forma `(2,)`:

`[hours_studied, exercises_solved]`

Uneori un model așteaptă un batch, chiar dacă avem un singur student.

În acest caz adăugăm o axă nouă:

- din forma `(2,)`;
- în forma `(1, 2)`.

În PyTorch, această operație este numită frecvent `unsqueeze`. În NumPy folosim `np.expand_dims`.

````python
single_student = X[0]

student_batch = np.expand_dims(single_student, axis=0)
student_column = np.expand_dims(single_student, axis=1)

print("Student:", single_student.shape)
print("Batch cu un student:", student_batch.shape)
print("Vector coloană:", student_column.shape)
````

Interpretare:

- `(2,)` - un vector cu două features;
- `(1, 2)` - un batch care conține un student;
- `(2, 1)` - un vector coloană.

### Schimbarea axelor: `np.swapaxes`

`np.swapaxes(array, axis1, axis2)` schimbă între ele două axe.

Pentru o matrice 2D, schimbarea axelor `0` și `1` transformă liniile în coloane și coloanele în linii.

````python
X_swapped = np.swapaxes(X, 0, 1)

print("Forma inițială:", X.shape)
print("Forma după swap:", X_swapped.shape)
print(X_swapped)
````

## 11. Produsul scalar

Avem un vector de features pentru un student:

`[hours_studied, exercises_solved]`

și un vector de ponderi:

`[importance_hours, importance_exercises]`

Produsul scalar combină cele două features într-un singur scor:

\[
score = hours\_studied \cdot importance\_hours
      + exercises\_solved \cdot importance\_exercises
\]

În NumPy putem calcula același rezultat cu `np.dot`.

````python
features = np.array([4, 18])
weights = np.array([0.7, 0.3])

manual_score = (
    features[0] * weights[0]
    + features[1] * weights[1]
)

numpy_score = np.dot(features, weights)

print("Calcul manual:", manual_score)
print("np.dot:", numpy_score)
````

## 12. Diferența dintre `*` și `@`

În cazul array-urilor NumPy, operatorii `*` și `@` au semnificații diferite.

### `*` - înmulțire element cu element

Fiecare element este înmulțit cu elementul aflat pe aceeași poziție.

Dacă formele sunt diferite, NumPy poate folosi **broadcasting**, atât timp cât formele sunt compatibile.

Broadcasting înseamnă că un array mai mic este aplicat automat peste un array mai mare, fără să îl copiem manual.

De exemplu, pentru o matrice cu forma `(4, 2)` și un vector cu forma `(2,)`, același vector este aplicat fiecărei linii.

### `@` - produs matriceal

`@` combină liniile primei matrici cu coloanele celei de-a doua matrici.

În Machine Learning, `@` este folosit pentru a combina datele de intrare cu ponderile unui model.

````python
student_features = np.array([
    [4.0, 18.0],
    [2.0, 10.0],
    [5.0, 25.0],
    [3.0, 14.0],
])

feature_weights = np.array([1.0, 0.5])

elementwise_result = student_features * feature_weights

print("Rezultat cu *:")
print(elementwise_result)
````

În exemplul de mai sus, broadcasting-ul aplică vectorul `[1.0, 0.5]` fiecărui student:

- orele studiate sunt înmulțite cu `1.0`;
- exercițiile rezolvate sunt înmulțite cu `0.5`.

NumPy compară dimensiunile de la dreapta la stânga.

Două dimensiuni sunt compatibile dacă:

- sunt egale;
- sau una dintre ele este `1`;
- sau una dintre dimensiuni lipsește.

## 13. Înmulțirea matricilor

Pentru produsul matriceal, dimensiunile interioare trebuie să coincidă.

Dacă:

- `A` are forma `(m, n)`;
- `B` are forma `(n, p)`;

atunci:

- `A @ B` are forma `(m, p)`.

În exemplul nostru:

- `X` are forma `(număr_studenți, număr_features)`;
- `W` are forma `(număr_features, număr_outputs)`;
- `X @ W` are forma `(număr_studenți, număr_outputs)`.

Pentru fiecare student, NumPy combină toate features cu fiecare coloană de ponderi și produce câte un scor de ieșire.

````python
X = np.array([
    [4.0, 18.0],
    [2.0, 10.0],
    [5.0, 25.0],
    [3.0, 14.0],
])

W = np.array([
    [0.7, -0.2],
    [0.3,  0.5],
])

scores = X @ W

print("X shape:", X.shape)
print("W shape:", W.shape)
print("Scores shape:", scores.shape)
print(scores)
````

## 14. Transformarea liniară: \(A = XW + b\)

O operație fundamentală în Machine Learning este:

\[
A = XW + b
\]

În cod NumPy:

`A = X @ W + b`

Unde:

- `X` conține datele de intrare;
- `W` conține ponderile;
- `b` este bias-ul;
- `A` conține scorurile produse de model.

Pentru un batch de studenți:

- `X`: `(batch_size, input_features)`;
- `W`: `(input_features, output_features)`;
- `b`: `(output_features,)`;
- `A`: `(batch_size, output_features)`.

Bias-ul este adăugat fiecărei linii prin broadcasting.

````python
X = np.array([
    [4.0, 18.0],
    [2.0, 10.0],
    [5.0, 25.0],
    [3.0, 14.0],
])

W = np.array([
    [0.7, -0.2],
    [0.3,  0.5],
])

b = np.array([0.1, -0.4])

A = X @ W + b

print("X shape:", X.shape)
print("W shape:", W.shape)
print("b shape:", b.shape)
print("A shape:", A.shape)
print(A)
````
